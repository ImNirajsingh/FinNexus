## Create merchant/category features

```python
def categorize_transaction(row):

    text = (
        str(row["description_clean"]) + " " +
        str(row["transaction_message"]) + " " +
        str(row["counterparty"])
    )

    if any(word in text for word in [
        "SWIGGY", "ZOMATO", "FOOD"
    ]):
        return "Food"

    if any(word in text for word in [
        "UBER", "OLA", "RAPIDO"
    ]):
        return "Transport"

    if any(word in text for word in [
        "AMAZON", "FLIPKART"
    ]):
        return "Shopping"

    if any(word in text for word in [
        "AIRTEL", "JIO", "RECHARGE"
    ]):
        return "Recharge"

    if any(word in text for word in [
        "RENT"
    ]):
        return "Rent"

    if any(word in text for word in [
        "SALARY", "PAYROLL"
    ]):
        return "Salary"

    return "Other"
```

```python
df['transaction_category'] = df.apply(categorize_transaction, axis=1)
```

## Improve categorization using transaction direction

```python
def categorize_transaction(row):

    text = str(row["description_clean"])

    if "SALARY" in text:
        return "Salary"

    if row["transaction_type"] == "CR":
        if "UPI" in text:
            return "Money Received"

    if row["transaction_type"] == "DR":
        if "RECHARGE" in text:
            return "Recharge"

    return "Other"
```

```python
df = df.sort_values("transaction_datetime").reset_index(drop=True)

df["previous_balance"] = df["balance"].shift(1)
```
